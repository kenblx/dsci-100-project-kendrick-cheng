#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
womens_march_madness_tidy <- womens_march_madness_csv |> # use original read csv as data frame in pipe operator
    filter(year > 2000) |> # filter for year past 2000
    select(year, school, seed, conf_wins, conf_losses, reg_wins, reg_losses, tourney_wins, tourney_losses, tourney_finish) |> # select variables we need or want
    drop_na(conf_wins, conf_losses, reg_wins, reg_losses) |> # drop non applicable values we need for calculation
    mutate(tourney_wins_pct = round(tourney_wins / tourney_losses * 100, 2)) |>
    mutate(conf_reg_wins_pct = round((conf_wins + reg_wins) / (conf_wins + reg_wins + conf_losses + reg_losses) * 100, 2)) 
# the mutate line creates the new combined conference and regional wins percentage column

head(womens_march_madness_tidy, 6) # show the first 6 rows of the tidied data
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
womens_march_madness_numeric <- womens_march_madness_tidy |>
    select(-school, -tourney_finish, -seed, -year)

womens_march_madness_mean <- womens_march_madness_numeric |>
    map_dfr(mean, na.rm = TRUE) |>
    mutate(measure = "mean")

womens_march_madness_min <- womens_march_madness_numeric |>
    map_dfr(min, na.rm = TRUE) |>
    mutate(measure = "min")

womens_march_madness_max <- womens_march_madness_numeric |>
    map_dfr(max, na.rm = TRUE) |>
    mutate(measure = "max")

womens_march_madness_info <- bind_rows(womens_march_madness_mean, womens_march_madness_min, womens_march_madness_max)
womens_march_madness_info
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).

```r
womens_march_madness_na <- womens_march_madness_tidy |>
    summarize(across(everything(), ~ sum(is.na(.))))
womens_march_madness_na
```

```r
wmm_school_reg_wins_mean <- womens_march_madness_tidy |>
    select(school, reg_wins) |>
    group_by(school) |>
    summarize(mean_reg_wins = round(mean(reg_wins), 2))
wmm_school_reg_wins_mean
```
